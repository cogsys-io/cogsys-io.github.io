### 2.3 Nowość: instrukcja warunkowa `if / elif / else`

Dotąd każdy nasz program robił zawsze to samo. `if` pozwala wybrać, co się wykona:

````python
liczba = 7
if liczba % 2 == 0:
    print("parzysta")
else:
    print("nieparzysta")
````

Zasady:

- po `if` jest **warunek** — coś, co daje `True` albo `False` — i **dwukropek**;
- blok pod spodem jest **wcięty** (4 spacje; Jupyter robi to sam po dwukropku) — wcięcie mówi Pythonowi, co należy do `if`;
- `else` łapie wszystko, co nie pasowało;
- `elif` („else if”) dodaje kolejne warunki. Python sprawdza je **po kolei od góry** i wykonuje **tylko pierwszy** pasujący blok.

````python
godzina = 14
if godzina < 12:
    print("Dzień dobry — przed południem")
elif godzina < 18:
    print("Dzień dobry — po południu")
else:
    print("Dobry wieczór")
````

Przypomnienie: warunki łączymy słowami `and`, `or`, `not`; porównania można „łańcuszkować”: `1 <= miesiac <= 12`.

### 2.4 Nowość: własne funkcje — `def`

Funkcja to **przepis z nazwą**: raz zapisany, wywoływany wiele razy z różnymi danymi. Gotowych używamy od początku (`print`, `len`, `int`, `datetime.date.today`). Własną definiujemy tak:

````python
def powitanie(imie):
    return f"Cześć, {imie}!"

print(powitanie("Ala"))
print(powitanie("Bartek"))
tekst = powitanie("Celina")     # wynik można zapisać do zmiennej...
print(tekst.upper())            # ...i dalej z nim pracować
````

`def` + nazwa + **parametry** w nawiasach + dwukropek; ciało wcięte; `return` **zwraca wynik** do miejsca wywołania.

Częsta pomyłka: `print` tylko *wyświetla*, nic nie zwraca. Funkcja bez `return` zwraca `None` („nic”):

````python
def powitanie_print(imie):
    print(f"Cześć, {imie}!")

wynik = powitanie_print("Ala")
print(wynik)          # None — funkcja nic nie zwróciła
````
