## 2. Ile dni żyjesz? Wersja „na piechotę” (10 min)

Zapytajmy o pełną datę urodzenia. Dziś jest 23 października 2026. Spróbujmy policzyć przeżyte dni samodzielnie, bez żadnej pomocy.

```python
dzien_ur = int(input("Dzień urodzin (1-31): "))
miesiac_ur = int(input("Miesiąc urodzin (1-12): "))
rok_ur = int(input("Rok urodzin: "))
print(f"Data urodzenia: {dzien_ur}.{miesiac_ur}.{rok_ur}")
```

```python
# Dzisiejsza data — wpisana ręcznie, bo na razie nie umiemy jej „pobrać”
dzien_dzis = 23
miesiac_dzis = 10
rok_dzis = 2026

# Pierwsze przybliżenie: pełne lata razy 365
dni = (rok_dzis - rok_ur) * 365
print(f"Mniej więcej {dni} dni")
```

Czego tu brakuje? Zastanów się chwilę, zanim czytasz dalej.

- **Lata przestępne:** co 4 lata luty ma 29 dni… ale rok podzielny przez 100 nie jest przestępny… chyba że dzieli się przez 400. (1900 — nie, 2000 — tak, 2100 — nie.)
- **Miesiące** mają 28, 29, 30 albo 31 dni — trzeba doliczyć dni od urodzin do dziś w obrębie roku.
- Jeśli urodziny w tym roku **jeszcze nie minęły**, pełnych lat jest o jeden mniej.

```python
# Reguła roku przestępnego jako wyrażenie logiczne (wynik: bool)
rok = rok_ur
przestepny = (rok % 4 == 0 and rok % 100 != 0) or (rok % 400 == 0)
print(f"Rok {rok} jest przestępny? {przestepny}")
```

```python
# Nieco lepsze przybliżenie: średni rok ma 365.25 dnia, średni miesiąc 30.44 dnia
dni = (rok_dzis - rok_ur) * 365.25 + (miesiac_dzis - miesiac_ur) * 30.44 + (dzien_dzis - dzien_ur)
print(f"Około {dni:.0f} dni")
```

To już „prawie dobrze”. Ale o ile dni się mylimy? Nie wiemy — i to jest problem. Żeby policzyć dokładnie, trzeba by obsłużyć długości wszystkich miesięcy, reguły lat przestępnych, przypadek „urodziny jeszcze nie minęły”… Da się, ale to kilkadziesiąt linii kodu, w których łatwo o błąd. (Excel do dziś twierdzi, że rok 1900 był przestępny — bo tak liczył program Lotus 1-2-3 z 1983 r., a Excel chciał być z nim zgodny.)

Ktoś już to zrobił — porządnie, dawno temu i za darmo. Wystarczy to zaimportować.

## 3. Moduł `datetime` (22 min)

**Moduł** to gotowy kod (plik albo katalog plików), który dołączamy do programu jednym słowem: `import`. Na poprzednich zajęciach pojawiły się już `math` (np. `math.sqrt`) i `os`. Razem z Pythonem dostajemy **bibliotekę standardową** — kilkaset modułów gotowych do użycia bez instalowania czegokolwiek. Jeden z nich to `datetime`.

```python
import datetime

dzisiaj = datetime.date.today()
print(dzisiaj)
print(type(dzisiaj))
```

`datetime.date.today()` czytamy od lewej: w module `datetime` jest typ `date`, a on ma funkcję `today()`. Wynik to nowy typ danych — `date` — ze swoimi własnymi „częściami”:

```python
print(dzisiaj.year)
print(dzisiaj.month)
print(dzisiaj.day)
print(dzisiaj.weekday())    # 0 = poniedziałek, ..., 6 = niedziela
```

```python
# Własną datę tworzymy, podając rok, miesiąc i dzień:
urodziny = datetime.date(rok_ur, miesiac_ur, dzien_ur)
print(urodziny)
```

```python
# Daty można od siebie odejmować!
roznica = dzisiaj - urodziny
print(roznica)
print(type(roznica))
```

Wynik odejmowania dwóch dat to kolejny typ: `timedelta` — „odcinek czasu”. Ma pole `.days`:

```python
print(f"Żyjesz już {roznica.days} dni.")
print(f"To około {roznica.days / 365.25:.1f} lat, czyli {roznica.days / 7:.0f} tygodni.")
```

Porównaj z wynikiem „na piechotę” z punktu 2. O ile dni się różnią?

Lata przestępne, długości miesięcy, „urodziny jeszcze nie minęły” — moduł załatwił to za nas. I to poprawnie dla każdej daty od roku 1 do roku 9999.

### Krócej: `fromisoformat`

Zamiast trzech pytań można zadać jedno — o datę w formacie ISO: `RRRR-MM-DD` (np. `2005-03-14`). Ten zapis rozumie cały świat, a Python potrafi go zamienić na `date`:

```python
tekst = input("Data urodzenia (RRRR-MM-DD): ")
urodziny = datetime.date.fromisoformat(tekst)
print(urodziny, type(urodziny))
print((dzisiaj - urodziny).days, "dni")
```

### Odcinki czasu można dodawać do dat

`timedelta` tworzymy też sami, podając liczbę dni (albo godzin, minut, sekund…):

```python
tydzien = datetime.timedelta(days=7)
print(dzisiaj + tydzien)          # za tydzień
print(dzisiaj - tydzien)          # tydzień temu
print(urodziny + datetime.timedelta(days=1000))   # Twój tysięczny dzień życia
```

### Nie tylko dni: `datetime.datetime`

Typ `date` to sama data. Typ `datetime` to data **i godzina** — tak, nazywa się tak samo jak moduł, stąd podwójne `datetime.datetime`:

```python
teraz = datetime.datetime.now()
print(teraz)
print(teraz.hour, teraz.minute, teraz.second)
```

```python
# Moment urodzenia — przyjmijmy północ, jeśli nie znasz godziny
narodziny = datetime.datetime(urodziny.year, urodziny.month, urodziny.day, 0, 0)
zycie = teraz - narodziny
print(zycie)
print(f"Żyjesz już {zycie.total_seconds():,.0f} sekund.")
```

## 4. Dzień, którego nie było — po co nam moduły? (10 min)

### Samoa, grudzień 2011

Samoa to wyspiarskie państwo na Pacyfiku (stolica: Apia), tuż przy **międzynarodowej linii zmiany daty**. Do 2011 r. leżało po jej wschodniej stronie — było jednym z ostatnich miejsc na Ziemi, gdzie kończył się każdy dzień. Od Nowej Zelandii i Australii, głównych partnerów handlowych (kilka godzin lotu), dzieliła je prawie cała doba: gdy w Apia zaczynał się piątek, w Auckland zaczynała się już sobota. Wspólnych dni roboczych w tygodniu były cztery.

Rząd Samoa postanowił więc „przeskoczyć” na zachodnią stronę linii. W **czwartek 29 grudnia 2011** mieszkańcy poszli spać, a obudzili się w **sobotę 31 grudnia**. Piątek 30 grudnia 2011 w Samoa nigdy nie nastąpił. (Nikt nie obchodził tego dnia urodzin, ale pracodawcy wypłacili pensje za cały tydzień.) Zresztą w 1892 r. Samoa zrobiło ruch w drugą stronę — żeby handlować z USA — i wtedy jeden dzień, 4 lipca 1892, przeżyło **dwa razy**.

Pytanie dla programisty: ile dni żyje ktoś urodzony w Apia? Czy Twój kalkulator „na piechotę” z punktu 2 wiedziałby o tym? Zobaczmy, czy wie o tym Python.

```python
from zoneinfo import ZoneInfo     # strefy czasowe — też w bibliotece standardowej

apia = ZoneInfo("Pacific/Apia")
utc = datetime.timezone.utc       # czas uniwersalny — wspólny „zegar” dla całej Ziemi

# Trzy kolejne południa UTC, 29–31 grudnia 2011...
p1 = datetime.datetime(2011, 12, 29, 12, 0, tzinfo=utc)
p2 = p1 + datetime.timedelta(days=1)
p3 = p2 + datetime.timedelta(days=1)

# ...i te same chwile widziane z Apia:
print(p1, "->", p1.astimezone(apia))
print(p2, "->", p2.astimezone(apia))
print(p3, "->", p3.astimezone(apia))
```

Spójrz na prawą kolumnę — kalendarz w Apia: 29, 31, 1. Trzydziestego nie ma. Końcówka `-10:00` (10 godzin za czasem uniwersalnym) zmieniła się w `+14:00` (14 godzin przed nim): skok o 24 godziny.

Czy pisząc własny kalkulator, moglibyśmy to przewidzieć? Nie. Czy ktoś to za nas sprawdził? Tak. Strefy czasowe całego świata — razem z całą ich historią — są opisane w jednej, publicznej bazie (*tz database*), poprawianej za każdym razem, gdy jakiś rząd zmieni zasady. A to zdarza się kilka razy w roku. Moduł `zoneinfo` po prostu z niej korzysta.

```python
# Ile czasu minęło w Apia od południa 29 XII do południa 31 XII 2011?
przed = datetime.datetime(2011, 12, 29, 12, 0, tzinfo=apia)
po    = datetime.datetime(2011, 12, 31, 12, 0, tzinfo=apia)

print("Kalendarz mówi:", (po.date() - przed.date()).days, "dni")
print("Zegar mówi:    ", po.astimezone(utc) - przed.astimezone(utc))
```

Pułapka (nawet z modułem trzeba czasem zajrzeć do dokumentacji): gdy odejmujemy dwa `datetime` z **tą samą** strefą, Python porównuje wskazania ściennego zegara, a nie rzeczywisty upływ czasu. Dlatego przed odejmowaniem przeliczamy obie chwile na UTC (`.astimezone(utc)`).

### Bliżej domu: najbliższa noc z soboty na niedzielę

Nie trzeba lecieć na Pacyfik. W nocy z 24 na 25 października 2026 w Polsce cofamy zegarki z 3:00 na 2:00. Ta noc ma 25 godzin:

```python
warszawa = ZoneInfo("Europe/Warsaw")
sobota    = datetime.datetime(2026, 10, 24, 12, 0, tzinfo=warszawa)
niedziela = datetime.datetime(2026, 10, 25, 12, 0, tzinfo=warszawa)

print("Kalendarz:", (niedziela.date() - sobota.date()).days, "dzień")
print("Zegar:    ", niedziela.astimezone(utc) - sobota.astimezone(utc))
print("Przesunięcie względem UTC:", sobota.utcoffset(), "->", niedziela.utcoffset())
```

### Ciekawostka: dni, których nie było w Polsce

Python zakłada, że kalendarz gregoriański obowiązywał „od zawsze”. W Polsce wprowadzono go w 1582 r.: po czwartku 4 października nastąpił od razu piątek 15 października. Python o tym nie wie — dla dat sprzed 15 X 1582 jego wyniki nie zgadzają się z historią:

```python
print(datetime.date(1582, 10, 10))   # w Polsce tego dnia nie było — Python nie protestuje
```

### Morał: po co moduły?

- **Ktoś już to rozwiązał** — lata przestępne, długości miesięcy, dni tygodnia, strefy czasowe. Jedna linia `import` zamiast kilkudziesięciu linii własnego kodu.
- **Kod z modułu jest sprawdzony** przez miliony użytkowników. Twoja własna wersja — jeszcze nie.
- **Świat się zmienia** (Samoa, zmiany czasu, nowe przepisy). Bazę stref ktoś aktualizuje; Twój kod nie musi się zmieniać.
- **Czytelność:** `dzisiaj - urodziny` mówi samo za siebie.
- Biblioteka standardowa to „baterie w zestawie”: `math`, `os`, `datetime`, `calendar`, `zoneinfo`, `random`, `json`, … Zanim coś napiszesz od zera, sprawdź, czy już istnieje.
- Ale: moduł ma założenia i granice (kalendarz „od zawsze” gregoriański, pułapka z odejmowaniem). Dokumentacja to nie wróg.
